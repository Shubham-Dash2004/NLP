```python
## why functions?
num=24
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if(num%2==0):
        print("even number")
    else :
        print("odd number")    
```

```python
## Call this function
even_or_odd(10)
```

```python
## function with multiple parameters

def add(a,b):
    return a+b

res = add(4,6)
print(res)
```

```python
## Default Parameters

def greet(name):
    print(f"Hello {name} Welcome To the paradise")

greet("Dash")
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome To the paradise")

greet() # Guest
greet("Dash") # Dash
```

```python
### Variable Length Arguments
## Positional And Keywords arguments

## Positional arguments
# *args is used to take any number of arguments 
def numbers(*args):  # its not mandatory to use args , you can use any name with asterisk(*) *dash , *argument , *xyz etc
    for num in args:
        print(num)

numbers(1,2,3,4,5,"Dash",True)

```

```python
### Keywords Arguments
def kv(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

kv(name="Dash",age=22,country='India')
```

```python
def both(*args , **kwargs):
    for num in args:
        print("arguments: ",num)

    for key,value in kwargs.items():
        print(f"keyword arguments= {key}:{value}")

both(1,2,3,4,5,"Dash",True,name="Dash",age=22,country="india") # first write all your positional argument then followed by keyword argument
```

```python
### Return statements
def sum(a,b):
    return a+b
sum(10,20)
```
