Classes and Objects
Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modeling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including instance variables and methods.

Inheritance	Child gets properties/methods from parent
Encapsulation	Keep data/implementation controlled inside a class
Abstraction	Show required functionality, hide implementation details
Polymorphism	Same method/interface, different behavior

```python
# self will be references to the object so suppose employee class is there and e is an object of employee class, then e is an self. 
# self is a way to reference the object of the class which is being created
# and you must specify self as the first argument in each and every method you define inside class 

class Employee:
    company = 'hp'

    def get_salary(self): 
        return 30000


e1 = Employee()
print(e1)
print(e1.company)
print(e1.get_salary())

e2 = Employee()
print(e2)
print(e2.company)
print(e2.get_salary())
```

```python
# __init__ is a special method that runs automatically when you create an object.
# self simply means : the current object
class Person:
    # by default whenever we write init , it means the constructor and constructor is used to initialize an object 
    def __init__(self, name, age):  # __init__(person1, "Dash", 34)
        self.name = name  # Put the value of name inside the current object's name. : self.name = person1.name
        self.age = age

person1 = Person("Dash", 34)

print(person1.name)
print(person1.age)
```

```python
class Employee:
    company = "Samsung"

    # constructor
    def __init__(self,name,salary,bond):
        self.name = name
        self.salary = salary
        self.bond = bond

    # methods
    def get_salary(self):
        return self.salary

    def getInfo(self):
        print(f"the name of this employee is {self.name} and his salary is {self.salary} , and bond year is {self.bond}")


e1 = Employee("Dash",7000,2)
print(e1.get_salary())
e1.getInfo()
print(e1)
print(e1.company)

print("--------------------------")

e2 = Employee("Shubham",10000,3)
print(e2.get_salary())
e2.getInfo()
print(e2)
print(e2.company)

```

```python
class Employee:
    #class attribute
    company = "Samsung"

    # constructor
    def __init__(self,name,salary,bond,company):
        # instance attribute
        self.name = name   
        self.salary = salary
        self.bond = bond
        self.company = company

    # methods
    def get_salary(self):
        return self.salary

    def getInfo(self):
        print(f"the name of this employee is {self.name} and his salary is {self.salary} , and bond year is {self.bond}")



e1 = Employee("dash",7000,2,"Dell")
print(e1.company) # will always print instance attribute whenever present , if not then print class attribute
print(Employee.company) # it will print the class attribute because we are using Employee class name .

# Object Introspection 
print(dir(e1)) # to get all the methods and attributes that an object has , you can use dir(objectname)
```

```python
# inheritance and polymorphism
# For every method call, Python searches the child's class first, then its parents.
class Animal:
    def __init__(self):
        print("Animal Constructor called")

    def speak(self,sound):
        print(f"Animal speaks {sound}")    

class dog(Animal):
    def __init__(self):
        print("Dog Constructor called")
    def speak(self,sound):
        print(f"Dog barks {sound}")

# if init method is not present in child class then it will go to parent class and call the init method of parent class.
# method overriding is a feature of OOPs where if a method is present in both parent and child class , 
# and if we call that method using child class object then it will call the method of child class and not parent class.
d1 = dog()        
d1.speak("woof woof")  # it will call the speak method of dog class because it is present in dog class and it will not go to parent class.
```

```python
# method overriding and operator overloading are the two types of polymorphism in python.

class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def sum(self,other):
        return Point(self.x + other.x , self.y + other.y)

    def print_point(self):
        print(f"the point is ({self.x},{self.y})")
    
    def __add__(self,other):  # operator overloading
        return Point(self.x + other.x , self.y + other.y)

p1 = Point(2,3)
p2 = Point(4,5)

# p3 = p1.sum(p2)
# p3.print_point()

p3 = p1 + p2  # operator overloading
p3.print_point()
```
