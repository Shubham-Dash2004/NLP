Encapsulation And Abstraction
Encapsulation and abstraction are two fundamental principles of Object-Oriented Programming (OOP) that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features.

Encapsulation
Encapsulation is the concept of wrapping data (variables) and methods (functions) together as a single unit. It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

private = you cannot access any private variable or method outside the class even not in child class
protected = you cannot access protected variable or method outside the class , but you access it in child class

```python
### Encapsulation  with Getter and Setter MEthods
### Public,protected,private variables or access modifiers

class Person:
    def __init__(self,name):
        self.name = name  # public variable or access modifier

def get_name(person):
    return person.name

person = Person("john")
get_name(person) # Output: john
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.__name = name # private variable or private access modifier
        self.__age = age  # private variable or private access modifier
        self.gender = gender # public variable or public access modifier

def get_name(person):
    return person.__name # you can access it using _Person__name , but it is not recommended to access private variables outside the class.

def get_age(person):
    return person.__age # you can access it using _Person__age , but it is not recommended to access private variables outside the class.
    

person = Person("john",20,"male")
get_name(person) # this will give an error because __name is a private variable and it cannot be accessed outside the class.
get_age(person) # this will give an error because __age is a private variable and it cannot be accessed outside the class.
```

```python
dir(person)  # only public variable will be shown in the output. Private variables will not be shown in the output. 
# only gender will be shown in the output. there will be no name and age in the output because they are private variables.
```

```python
class Person:
    def __init__(self,name,age):
        self._name = name  # protected variable or access modifier
        self._age = age  # protected variable or access modifier

# def get_name(person):
#     return person._name # you can access this protected variable outside the class but it is not recommended to access protected variables outside the class.
# # python trust the developer that they will not access the protected variable outside the class.

class Employee(Person):
    def __init__(self,name,age):
        super().__init__(name,age)

# person = Person("john",20)
# print(person._name) 
# print(person._age)

e1 = Employee("john",20)
print(e1._name) # this will work because we are accessing the protected variable inside the subclass. it is recommended to access protected variables inside the subclass.
```

```python
# "_name and _age are protected-by-convention attributes. They are intended to be accessed within the class and its subclasses, although Python still allows outside access."
class Person:
    def __init__(self,name,age):
        self._name = name  # protected variable or access modifier
        self._age = age  # protected variable or access modifier

class Employee(Person):
    def __init__(self,name,age):
        super().__init__(name,age)

    def show_details(self):
        print(f"Name: {self._name}, Age: {self._age}")    


e1 = Employee("john",20)
e1.show_details()
```

```python
## Encapsulation With Getter And Setter
class Person:
    def __init__(self,name,age):
        self.__name=name  ## Private access modifier or variable
        self.__age=age ## Private variable

    ## getter method for name
    def get_name(self):
        return self.__name
    
    ## setter method for name
    def set_name(self,name):
        self.__name=name

    # Getter method for age
    def get_age(self):
        return self.__age
    
    # Setter method for age
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age cannot be negative.")


person=Person("Dash",34)

## Access and modify private variables using getter and setter

print(person.get_name())
print(person.get_age())

person.set_age(35)
print(person.get_age())
person.set_age(-5)
```
