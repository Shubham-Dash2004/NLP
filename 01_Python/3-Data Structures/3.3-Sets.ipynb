Sets
Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order (we cant access element by index like we do in list and tuple ), and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

A Set in python is an unordered , mutable and unique collection of elements . it does not allow duplicate values .

```python
empty_set = set()
print(type(empty_set))  # Output: <class 'set'>
# remeber one thing that if you want to create an empty set, you have to use set() and not {} because {} creates an empty dictionary.
```

```python
empty_set = {}
print(type(empty_set))  # Output: <class 'dict'>
# you have to use set() to create an empty set, not {} because {} creates an empty dictionary.
```

```python
my_set = {1,2,3,4,5}
print(my_set)  # Output: {1, 2, 3, 4, 5}
print(type(my_set))  # Output: <class 'set'>
```

```python
# you can also put list and other data strucutre in set but remember it will filter out non unique elements , no order is followed , and can be changed (mutable)
#list to set 
my_set = set([1,2,3,4,5])
print(my_set)  # Output: {1, 2, 3, 4, 5}
```

```python
my_empty_set = set([1,2,3,4,5,1,2,3])
print(my_empty_set)  # Output: {1, 2, 3, 4, 5} - duplicates are removed
```

```python
## Basics Sets Operation
## Adding and Removing Elements
my_set.add(6)
print(my_set)  # Output: {1, 2, 3, 4, 5, 6} 
my_set.add(6)
print(my_set)  # Output: {1, 2, 3, 4, 5, 6} - duplicates are not allowed
```

```python
## Remove the elements from a set
my_set.remove(6)
print(my_set)  # Output: {1, 2, 3, 4, 5}
```

```python
my_set.remove(10)  # This will raise a KeyError because 10 is not in the set
```

```python
my_set.discard(11)  # This will not raise an error even if 11 is not in the set
print(my_set)  # Output: {1, 2, 3, 4, 5}
```

```python
my_set.discard(5)
print(my_set)  # Output: {1, 2, 3, 4}
```

```python
removed_element = my_set.pop()  # Removes and returns an arbitrary element from the set # that element can be any element from the set, as sets are unordered collections.
print(removed_element)  # Output: An arbitrary element from the set 
print(my_set)  # Output: The set with the removed element
```

```python
## clear all the elements
my_set.clear()
print(my_set)
```

```python
## Set Memebership test
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # Output: True
print(10 in my_set)  # Output: False
```

```python
## Mathematical Operation
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

## Union of two sets
union_set = set1.union(set2)
print(union_set)  # Output: {1, 2, 3, 4, 5, 6, 7, 8}

## Intersection of two sets
intersection_set = set1.intersection(set2)
print(intersection_set)  # Output: {4, 5}

set1.intersection_update(set2)
print(set1)  # Output: {4, 5} - set1 is updated to the intersection of set1 and set2
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

print(set1.difference(set2))  # Output: {1, 2, 3} - elements in set1 but not in set2
```

```python
set1
```

```python
set2.difference(set1) # Output: {7, 8, 9} - elements in set2 but not in set1
```

```python
## Symmetric Difference = set1.difference(set2) + set2.difference(set1) 
set1.symmetric_difference(set2)  # Output: {1, 2, 3, 7, 8, 9} - elements in either set1 or set2 but not in both
```

```python
## Sets Methods
set1={1,2,3,4,5}
set2={3,4,5}
# is subset
print(set1.issubset(set2))  # Output: False - set1 is not a subset of set2
print(set2.issubset(set1))  # Output: True - set2 is a subset of set1
print(set1.issuperset(set2))  # Output: True - set1 is a superset of set2
print(set2.issuperset(set1))  # Output: False - set2 is not a superset of set1
```

```python
lst = [1,2,3,4,5]
set(lst)  # Output: {1, 2, 3, 4, 5} - converts list to set
```

```python
### Counting Unique words in text

text="In this tutorial we are discussing about sets"
words=text.split()  # Because we didn't tell Python what to split by, Python uses spaces by default.

## convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```

```python
#mixed data type
mixed_set = {1, "Hello", 3.14, True}
print(mixed_set)  # Output: {1, 'Hello', 3.14, True} - set can contain mixed data types
```

```python
s= {1, 2, 3, 4, 5}
s.update({4, 7, 8})
print(s)  # Output: {1, 2, 3, 4, 5, 7, 8} - set is updated with new elements
```

```python
se = {1, 2, 3, 4, 5}
set_copy = se.copy()
print(set_copy)  # Output: {1, 2, 3, 4, 5, 7, 8} - creates a shallow copy of the set
```


Conclusion
Sets are a powerful and flexible data type in Python that provide a way to store collections of unique elements. They support various operations such as union, intersection, difference, and symmetric difference, which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean Python code, especially when dealing with unique collections and membership tests.
