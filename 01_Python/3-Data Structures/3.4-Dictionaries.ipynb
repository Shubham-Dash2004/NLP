Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.


```python
empty_dict = {}
print(type(empty_dict))  # Output: <class 'dict'> - empty_dict is a dictionary
```

```python
empty_dict = dict()
print(type(empty_dict))  # Output: <class 'dict'> - empty_dict is a dictionary
```

```python
student = {"name" : "John" , "age" : 22 , "grade" : "A"}
print(student)  # Output: {'name': 'John', 'age': 22, 'grade': 'A'} - student is a dictionary with key-value pairs
print(type(student))  # Output: <class 'dict'> - student is a dictionary
```

```python
# duplicate keys are not allowed in dictionaries. If a key is repeated, the last value is retained.
student = {"name" : "John" , "age" : 22 , "name" : 24}
print(student)  # Output: {'name': 'A', 'age': 22} - duplicate keys are not allowed, the last value is retained
```

```python
## accessing Dictionary Elements
student={"name":"Roman","age":32,"grade":'A'}
print(student)
```

```python
print(student['age'])  # Output: 32 - accessing the value of the key 'age'
print(student['name'])  # Output: Roman - accessing the value of the key 'name'

# accessing the value using get method 
print(student.get('age'))
print(student.get('name'))
print(student.get('rollno')) # if no key with this name is there in dict then it will return None
print(student.get('rollno','Not_available'))  # if you want to custom return statement when key not found in dict use second argument 
```

```python
## Modifying Dicitonary Elements
# key = immutable , value = mutable
## Dictionary are mutable,so you can add, update or delete elements 
print(student)
```

```python
# update the value 
student['age'] = 22
print(student)
#add the value
student['country'] = 'India'
print(student)
```

```python
del student['country']
print(student)
```

```python
keys = student.keys()
print(key)
values = student.values()
print(values)
items = student.items()
print(items)
```

```python
# student_copy will not be a copy of student , it will point to the same address as the student does , so changes in student will reflect in student_copy \
# it will refer to student address 
student_copy = student
print(student_copy)
```

```python
# now age 30 is reflected in student_copy also because both variable points to same address so change in one will reflect in other 
# to prevent from this we have shallow copy
student['age'] = 30
print(student)
print(student_copy)
```

```python
student_copy = student.copy()
student['age'] = 35
print(student_copy)
print(student)
```

```python
### Iterating Over Dictionaries
## You can use loops to iterate over dictionatries, keys,values,or items

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterating over values
for values in student.values():
    print(values)
```

```python
## Iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
# nested dictonaries
students = {
    "student1": {"name":"Pat","age":35},
    "student2": {"name":"Starc","age":37}
}
```

```python
#accessing dictonaries
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
# dictionary comphrension 
squ = {x:x**2 for x in range(5)}
print(squ)
```

```python
# dictionary comphrension  with condition 
squ = {x:x**2 for x in range(10) if x%2==0} 
print(squ)
```

```python
# use a dictonary to count the frequency of elements  in list 

numbers = [1,2,2,3,3,3,4,4,4,4]
freq = {}
for num in numbers:
    if num in freq:
        freq[num] += 1
    else: 
       freq[num] = 1

print(freq)   
```

```python
## Merge 2 dictionaries into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
