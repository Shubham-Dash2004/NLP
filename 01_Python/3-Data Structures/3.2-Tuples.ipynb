Introduction to Tuples

Explanation:
Tuples are ordered collections of items that are immutable. They are similar to lists, but their immutability makes them different.

```python
## creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
# convert list to tuple 
tup = tuple([1,2,3,4])
print(type(tup))
print(tup)
```

```python
# convert tuple to list
lst = list((1,2,3,4))
print(type(lst))
print(lst)
```

```python
mixed_tuple=(1,"Hello World",3.14, True)
print(mixed_tuple)
```

```python
numbers = tuple([1,2,3,4])
print(numbers)
```

```python
print(numbers[-1])
print(numbers[2])
```

```python
numbers[-1:-3:-1]  # start = -1 , stop = -3, step = -1
```

```python
numbers[1::-1]
```

```python
numbers[0:4]
```

```python
## Tuple Operations

concatenation_tuple=numbers + mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple * 3
```

```python
numbers * 3
```

```python
## Immutable Nature Of Tuples
## Tuples are immutable, meaning their elements cannot be changed once assigned.

lst = [1, 2, 3]
lst[1] = 4  # This is valid, as lists are mutable
print(lst)
```

```python
tpl = (1, 2, 3)
tpl[1] = 4  # This will raise an error, as tuples are immutable
# if you want to change the value of tuple then you need to convert it into list and then change the value and then convert it back to tuple
print(tpl)
```

```python
numbers = (1, 2, 3, 4)
```

```python
# Tuple Methods
print(numbers.count(4))  # Output: 1
print(numbers.index(4))  # Output: 3
```

```python
## Packing and Unpacking tuple
## packing
packed_tuple = 1,"hello",3.14
print(packed_tuple)
```

```python
#unpacking a tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# unpacking a tuple with asterisk *
numbers = (1, 2, 3, 4, 5)
first , *middle, last = numbers
print(first)   # Output: 1
print(middle)  # Output: [2, 3, 4]
print(last)    # Output: 5
```

```python
## Nested Tuple
## Nested List
lst=[[1,2,3,4],[6,7,8,9],[1,"Hello",3.14,"c"]]
print(lst[0][2])
print(lst[2][1])
print(lst[1][0:4])
```

```python
# Nested List And Tuple
lst=[[1,2,3,4],[6,7,8,9],(1,"Hello",3.14,"c")]
print(lst[2][0:4])
```

```python
# Nested Tuple
nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))

## access the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
print(nested_tuple[2][0:2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()  # for new line after each sub-tuple    
```

Conclusion
Tuples are versatile and useful in many real-world scenarios where an immutable and ordered collection of items is required. They are commonly used in data structures, function arguments and return values, and as dictionary keys. Understanding how to leverage tuples effectively can improve the efficiency and readability of your Python code.
