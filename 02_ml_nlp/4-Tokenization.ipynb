```python
# NLTK is looking for tokenizer data , and it is not present in the newer version of nltk therefore we need to download the tokenizer data using the following command
import nltk

nltk.download('punkt_tab')  # this line will download the tokenizer data for NLTK. You can also download other data using nltk.download() function.
# nltk.download()  # this command will only work if your python version is 3.12.0 , otherwise it will not work 
```

```python
corpus="""Hello Welcome,to Dash's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## Paragraph-->Sentences
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)  # to execute this line first we need to download the tokenizer data for NLTK using the command nltk.download('punkt_tab')
```

```python
print(type(documents))
print(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
```

```python
## Paragraph-->words
words = word_tokenize(corpus)
print(words)  
```

```python
## sentence--->words
# you cant directly pass documents to word_tokenize() function because it is a list of sentences and word_tokenize() function only accepts string as input 
# therefore we need to iterate over the list of sentences and pass each sentence to the word_tokenize() function.
for sentence in documents:
    words = word_tokenize(sentence)
    print(words)
```

```python
from nltk.tokenize import wordpunct_tokenize 
```

```python
words = wordpunct_tokenize(corpus)  # wordpunct_tokenize() function is similar to word_tokenize() function but it also splits the punctuation from the words.
print(words) # it will treat punctuation as separate tokens and will not include them in the words.
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer() # it is used to add full stop with word not as seperate word but the last full stop will be removed from the word and 
# will be added to the next word as a separate token.
words = tokenizer.tokenize(corpus)
print(words)
```
